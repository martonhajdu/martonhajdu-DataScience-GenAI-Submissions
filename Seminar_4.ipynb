```python
import pandas as pd
df = pd.read_csv("students_two.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
df.isnull().sum()
```

```python
df = df.drop(['Id'], axis = 1)
df.head()
```

```python
# Converting features into numbers, so we can use all data for modeling
df = pd.get_dummies(df, columns=['High_School_Type'], drop_first=True, dtype=int)
df.head()

# Converted high school type into Private and State, we don't need a third column for other, if both are 0,
# we know it's in the other school type
```

```python
# Converting Scholarship to a number
df['Scholarship'] = df['Scholarship'].str.replace('%', '').astype(float) / 100
df.head()
```

```python
# Converting 'Additional work'
df['Additional_Work'] = df['Additional_Work'].map({'Yes': 1, 'No': 0})
df.head()
```

```python
# Checking for the values of these columns
categorical_cols = ['Sports_activity', 'Transportation', 'Attendance', 'Reading', 'Notes', 'Listening_in_Class', 'Project_work', 'Grade']

for col in categorical_cols:
    print(f"Unique values for '{col}': {df[col].unique()}")
```

```python
# Dropping the rows where 6 appears in 'Notes' and 'Listening_in_Class' and dropping the 3 from 'Attendance'
df = df[(df['Notes'] != '6') & (df['Listening_in_Class'] != '6') & (df['Attendance'] != '3')]
df.head()
```

```python
# Convertting 'Yes' and 'No' to 1 and 0
binary_cols = ['Sports_activity', 'Reading', 'Notes', 'Listening_in_Class', 'Project_work']

for col in binary_cols:
    df[col] = df[col].map({'Yes': 1, 'No': 0})

df.head()
```

```python
# Same for 'Transportation'
df['Transportation'] = df['Transportation'].map({'Private': 1, 'Bus': 0})
df.head()
```

```python
# For 'Attendance' a range makes sense
df['Attendance'] = df['Attendance'].map({'Never': 0, 'Sometimes': 1, 'Always': 2})
df.head()
```

```python
# Converting the 'Grade' column. Fail will be 0, rest (Pass) will be 1
df['Grade'] = df['Grade'].apply(lambda x: 0 if x == 'Fail' or 'D' in x else 1)
df.head()
```

```python
Y_values = df[['Grade']].values.ravel() # change to list
Y_values
```

```python
# Dropping 'Grade'. We will try to predict it, so it doesn't need to be part of the training data
X_values = df.drop(['Grade'], axis=1)
X_values.head()
```

```python
# Scalign X
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
# fit and transform the data and convert back to DataFrame
X_values = pd.DataFrame(scaler.fit_transform(X_values), columns=X_values.columns)
X_values
```

```python
# Training Test split
from sklearn.model_selection  import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X_values, Y_values, test_size = 0.2, random_state=1984, stratify=Y_values)

# print the shapes to check everything is OK
print(X_train.shape)
print(X_test.shape)
print(Y_train.shape)
print(Y_test.shape)
```

```python
# This is a classification problem
# We will use Logistic Regression for that
from sklearn.linear_model import LogisticRegression

# Create the Logistic Regression model
model = LogisticRegression()

# Train the model using the training data
model.fit(X_train, Y_train)
```

```python
# Make predictions on the test data
Y_pred = model.predict(X_test)

# Print the predictions
print(Y_pred)
```

```python
from sklearn.metrics import accuracy_score

# Calculate the accuracy of the model
accuracy = accuracy_score(Y_test, Y_pred)

# Print the accuracy
print(f"Accuracy: {accuracy}")
```

```python
from sklearn.tree import DecisionTreeClassifier

# Create the Decision Tree model
model_dt = DecisionTreeClassifier()

# Train the model using the training data
model_dt.fit(X_train, Y_train)
```

```python
Y_pred_dt = model_dt.predict(X_test)

from sklearn.metrics import accuracy_score

accuracy_dt = accuracy_score(Y_test, Y_pred_dt)

print(f"Decision Tree Accuracy: {accuracy_dt}")
```

```python
from sklearn.ensemble import GradientBoostingClassifier

# Instantiate a Gradient Boosting Classifier object
model_gb = GradientBoostingClassifier()

# Train the Gradient Boosting model
model_gb.fit(X_train, Y_train)
```

```python
# Make predictions on the test data
Y_pred_gb = model_gb.predict(X_test)

# Calculate the accuracy of the Gradient Boosting model
accuracy_gb = accuracy_score(Y_test, Y_pred_gb)

# Print the accuracy
print(f"Gradient Boosting Accuracy: {accuracy_gb}")
```

```python
print(f"Logistic Regression Accuracy: {accuracy}")
print(f"Decision Tree Accuracy: {accuracy_dt}")
print(f"Gradient Boosting Accuracy: {accuracy_gb}")

# Identify the best performing model
model_accuracies = {"Logistic Regression": accuracy, "Decision Tree": accuracy_dt, "Gradient Boosting": accuracy_gb}
best_model = max(model_accuracies, key=model_accuracies.get)
print(f"\nBased on accuracy, the best performing model is: {best_model}")
```

## Summary:

### Data Analysis Key Findings

*   The Decision Tree model achieved an accuracy of approximately 0.7143 on the test dataset.
*   The Gradient Boosting model achieved an accuracy of approximately 0.6786 on the test dataset.
*   Based on accuracy, both Logistic Regression and Decision Tree models performed equally well with an accuracy of approximately 0.714, which is better than the Gradient Boosting model's accuracy of approximately 0.679.

### Insights or Next Steps

*   Since Logistic Regression and Decision Tree models performed similarly in terms of accuracy, further evaluation using other metrics (e.g., precision, recall, F1-score) could provide more insights into their performance characteristics, especially if the dataset is imbalanced.
*   Hyperparameter tuning for each model, especially for Gradient Boosting which can be sensitive to its parameters, might lead to improved performance and potentially change the ranking of the models.


```python
from sklearn.model_selection import ParameterGrid

param_grid_dt = {
    'max_depth': [None, 5, 10, 15],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

param_grid_gb = {
    'n_estimators': [50, 100, 200],
    'learning_rate': [0.01, 0.1, 0.2],
    'max_depth': [3, 5, 7]
}
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeClassifier

# Instantiate a DecisionTreeClassifier object
dt = DecisionTreeClassifier()

# Instantiate GridSearchCV with the Decision Tree classifier and the parameter grid
grid_search_dt = GridSearchCV(dt, param_grid_dt, cv=5)

# Fit the GridSearchCV object to the training data
grid_search_dt.fit(X_train, Y_train)

# Print the best parameters found by GridSearchCV
print("Best parameters for Decision Tree:", grid_search_dt.best_params_)
```

```python
from sklearn.metrics import accuracy_score

# Get the best Decision Tree model from the grid search
best_dt_model = grid_search_dt.best_estimator_

# Make predictions on the test data using the best model
Y_pred_tuned_dt = best_dt_model.predict(X_test)

# Calculate the accuracy of the tuned Decision Tree model
accuracy_tuned_dt = accuracy_score(Y_test, Y_pred_tuned_dt)

# Print the accuracy
print(f"Tuned Decision Tree Accuracy: {accuracy_tuned_dt}")
```

```python
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import GridSearchCV

# Instantiate a GradientBoostingClassifier object
gb = GradientBoostingClassifier()

# Instantiate GridSearchCV with the Gradient Boosting classifier and the parameter grid
grid_search_gb = GridSearchCV(gb, param_grid_gb, cv=5)

# Fit the GridSearchCV object to the training data
grid_search_gb.fit(X_train, Y_train)

# Print the best parameters found by GridSearchCV
print("Best parameters for Gradient Boosting:", grid_search_gb.best_params_)
```

```python
# Get the best Gradient Boosting model from the grid search
best_gb_model = grid_search_gb.best_estimator_

# Make predictions on the test data using the best model
Y_pred_tuned_gb = best_gb_model.predict(X_test)

# Calculate the accuracy of the tuned Gradient Boosting model
accuracy_tuned_gb = accuracy_score(Y_test, Y_pred_tuned_gb)

# Print the accuracy
print(f"Tuned Gradient Boosting Accuracy: {accuracy_tuned_gb}")
```

```python
# Create a dictionary to store the accuracy of each model
model_accuracies_tuned = {
    "Logistic Regression": accuracy,
    "Tuned Decision Tree": accuracy_tuned_dt,
    "Tuned Gradient Boosting": accuracy_tuned_gb
}

# Print the accuracy of each model from the dictionary
print("Model Accuracies (After Tuning):")
for model_name, acc in model_accuracies_tuned.items():
    print(f"{model_name}: {acc}")

# Identify and print the name of the model with the highest accuracy
best_model_tuned = max(model_accuracies_tuned, key=model_accuracies_tuned.get)
print(f"\nBased on accuracy after tuning, the best performing model is: {best_model_tuned}")
```

## Summary:

### Data Analysis Key Findings

*   The best hyperparameters found for the Decision Tree model using GridSearchCV were `{'max_depth': None, 'min_samples_leaf': 4, 'min_samples_split': 10}`.
*   The accuracy of the tuned Decision Tree model on the test set is approximately 0.643.
*   The best hyperparameters found for the Gradient Boosting model using GridSearchCV were `{'learning_rate': 0.01, 'max_depth': 5, 'n_estimators': 100}`.
*   The accuracy of the tuned Gradient Boosting model on the test set is approximately 0.679.
*   Comparing the initial Logistic Regression model (accuracy: 0.714), the tuned Decision Tree model (accuracy: 0.643), and the tuned Gradient Boosting model (accuracy: 0.679), the Logistic Regression model achieved the highest accuracy.

### Insights or Next Steps

*   Despite hyperparameter tuning, neither the Decision Tree nor the Gradient Boosting models surpassed the performance of the initial Logistic Regression model on this dataset.
*   Further investigation into feature engineering or exploring other models might be beneficial to potentially improve performance beyond the Logistic Regression baseline.
